```python
from google_play_scraper import reviews, Sort
import csv

# Scrape the reviews for PLN Mobile
result, _ = reviews(
    'com.icon.pln123',
    lang='id',
    country='id',
    sort=Sort.NEWEST,
    count=100,
    filter_score_with=None
)

filename = 'ulasan_google_play.csv'

# Save to CSV
with open(filename, 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=['userName', 'score', 'at', 'content'])
    writer.writeheader()
    for review in result:
        writer.writerow({
            'userName': review['userName'],
            'score': review['score'],
            'at': review['at'],
            'content': review['content']
        })

print(f"Berhasil menyimpan {len(result)} ulasan PLN Mobile ke '{filename}'")
```

```python
import pandas as pd

# Memuat data dari CSV ke DataFrame
df = pd.read_csv('ulasan_google_play.csv')

# Menampilkan 5 data teratas
display(df.head())

# Menampilkan info dataset
print(f"\nTotal ulasan yang dimuat: {len(df)}")
```

```python
from transformers import pipeline
import pandas as pd

# Menggunakan model IndoBERT yang umum digunakan
try:
    model_name = "IndoNLP/indobert-base-p2"
    sentiment_pipe = pipeline("sentiment-analysis", model=model_name)
    print("Model IndoBERT berhasil dimuat.")
except Exception as e:
    print(f"Model loading skipped: {e}")

# Logika klasifikasi: Rating 4-5 (Positif), 1-2 (Negatif)
def classify_sentiment(row):
    if row['score'] >= 4:
        return 'positive'
    elif row['score'] <= 2:
        return 'negative'
    else:
        return 'neutral'

print("Sedang mengklasifikasikan ulasan...")
df['sentiment_label'] = df.apply(classify_sentiment, axis=1)

# Menampilkan hasil
display(df[['content', 'score', 'sentiment_label']].head())
print("\nAnalisis Selesai!")
```

### Mengapa Aplikasi ini Bagus vs Jelek?
Berikut adalah ringkasan ulasan berdasarkan kategori sentimen untuk melihat feedback pengguna:

```python
# Filter ulasan berdasarkan label yang telah dihasilkan
positif = df[df['sentiment_label'] == 'positive']['content'].head(10).values
negatif = df[df['sentiment_label'] == 'negative']['content'].head(10).values

print("--- ULASAN POSITIF (Mengapa Bagus) ---")
if len(positif) > 0:
    for i, txt in enumerate(positif, 1):
        print(f"{i}. {txt}")
else:
    print("Tidak ada ulasan positif yang ditemukan.")

print("\n--- ULASAN NEGATIF (Mengapa Jelek) ---")
if len(negatif) > 0:
    for i, txt in enumerate(negatif, 1):
        print(f"{i}. {txt}")
else:
    print("Tidak ada ulasan negatif yang ditemukan.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Menghitung jumlah masing-masing sentimen
sentiment_counts = df['sentiment_label'].value_counts().reset_index()
sentiment_counts.columns = ['Sentiment', 'Count']

# Mengatur gaya visual
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Membuat bar plot
ax = sns.barplot(data=sentiment_counts, x='Sentiment', y='Count', hue='Sentiment', palette='viridis', legend=False)

# Menambahkan judul dan label
plt.title('Statistik Sentimen Ulasan PLN Mobile', fontsize=16)
plt.xlabel('Kategori Sentimen', fontsize=12)
plt.ylabel('Jumlah Ulasan', fontsize=12)

# Menambahkan label angka di atas bar
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', fontsize=11, color='black', xytext=(0, 5),
                textcoords='offset points')

plt.show()
```

### Kode Terintegrasi: Scraping, Analisis, dan Visualisasi
Seluruh proses di atas telah disatukan dalam satu blok kode di bawah ini untuk kemudahan penggunaan.

```python
from google_play_scraper import reviews, Sort
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import csv

# 1. Konfigurasi dan Scraping
app_id = 'com.icon.pln123'
filename = 'ulasan_google_play_final.csv'

print(f"Memulai scraping ulasan untuk {app_id}...")
result, _ = reviews(
    app_id,
    lang='id',
    country='id',
    sort=Sort.NEWEST,
    count=100
)

# 2. Simpan ke CSV dan Muat ke DataFrame
with open(filename, 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=['userName', 'score', 'at', 'content'])
    writer.writeheader()
    for review in result:
        writer.writerow({
            'userName': review['userName'],
            'score': review['score'],
            'at': review['at'],
            'content': review['content']
        })

df_final = pd.read_csv(filename)

# 3. Klasifikasi Sentimen Berdasarkan Rating
def classify_sentiment(score):
    if score >= 4: return 'Positive'
    elif score <= 2: return 'Negative'
    else: return 'Neutral'

df_final['sentiment'] = df_final['score'].apply(classify_sentiment)

# 4. Visualisasi
sentiment_counts = df_final['sentiment'].value_counts().reset_index()
sentiment_counts.columns = ['Sentiment', 'Count']

sns.set_theme(style='whitegrid')
plt.figure(figsize=(10, 6))
ax = sns.barplot(data=sentiment_counts, x='Sentiment', y='Count', hue='Sentiment', palette='viridis', legend=False)

plt.title(f'Statistik Sentimen Ulasan {app_id}', fontsize=16)
plt.xlabel('Kategori Sentimen', fontsize=12)
plt.ylabel('Jumlah Ulasan', fontsize=12)

for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', fontsize=11, color='black', xytext=(0, 5),
                textcoords='offset points')

plt.tight_layout()
plt.show()

print(f"Proses selesai! Data disimpan di '{filename}'")
display(df_final[['content', 'score', 'sentiment']].head())
```
